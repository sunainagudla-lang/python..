```python
#dictionaries in python 
#Dictionary is a collection of key-values pairs that is unordered and mutable
student = {
    "name": "sunaina",
    "age": 00,
    "course": "python"
}

print(student)
```

```python
#access element in dic
student = {
    "name": "sunaina",
    "age": 00,
    "course": "python"

}

print(student["name"])
print(student["age"])
print(student["course"])
```

```python
#add new data to a dictionary
student["city"] = "vijayawada"

print(student)
```

```python
#keys, values, item, get, update, pop
student = {
    "name": "sunaina",
    "age":"19",
    "course": "python"

}

print(student.keys())
#keys()returns all keys

print(student)

print(student.values())
#values()returns all values

print(student)

print(student.items())
#items() returns all the values in the dictionary

print(student)

print(student.get("name"))
#get() returns the value of the specified key

print(student)

(student.update({"age":20}))
#updata() returns the value of specified key

print(student)

student.pop("age")
#pop() removes the specified keys and values

print(student)
```

```python
#popitem method
#popitem() removes the last inserted key-value pair
student = {
    "name": "sunaina",
    "age": 18,
    "course": "python"

}

student.popitem()

print(student)
```

```python
#setdefault method
student = {
    "name": "sunaina"
}
student.setdefault("age",21)

print(student)
```
